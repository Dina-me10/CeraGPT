```python
# [2]
text = 'My skin is very dry in winter so I need a good moisturizer and a gentle cleanser'  # 17
print(text)
```

```python
# [3] Tokenization: gapni so'zlarga bo'lish
tokens = text.split()
print(tokens)
```

```python
# [5] Vocabulary: har bir yangi so'zga ID berish
vocab = {}
current_id = 0

for token in tokens:
    if token not in vocab:
        vocab[token] = current_id
        current_id += 1

print(vocab)
```

```python
# [6] Encode: so'zlar → ID'lar
encoded = [vocab[token] for token in tokens]
print(encoded)
```

```python
# [8] Decode: ID'lar → so'zlar
reverse_vocab = {id: token for token, id in vocab.items()}
decoded = " ".join([reverse_vocab[id] for id in encoded])
print(decoded)
```

```python
# Embedding
# "moisturizer" => [0.41, 0.22, -0.77, 0.56, .........]
# "cream"       => [0.39, 0.25, -0.71, 0.60, .........]   ← ma'nosi yaqin → sonlar ham yaqin
# "laptop"      => [-0.88, 0.05, 0.43, -0.31, .........]  ← ma'nosi uzoq → sonlar butunlay boshqa
```

```python
import torch
import torch.nn as nn

embedding_dim = 4           # har bir token 4 ta sonli vektorga aylanadi
vocab_size = len(vocab)     # 16 ta noyob so'z

embedding_layer = nn.Embedding(vocab_size, embedding_dim)

input_ids = torch.tensor([encoded])   # [[0, 1, 2, ..., 10, 14, 15]]
embeddings = embedding_layer(input_ids)

print(embeddings)
print(embeddings.shape)
```

```python
from transformers import AutoTokenizer

model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
print("Vocab size:", tokenizer.vocab_size)
print("Special tokens:", tokenizer.special_tokens_map)
```

```python
# [13]
text = "Hello, which moisturizer is good for dry skin?"
print('text:', text)
```

```python
# [14]
tokens = tokenizer.tokenize(text)
print('tokens:', tokens)
```

```python
# [15]
ids = tokenizer.convert_tokens_to_ids(tokens)
print('token IDs:', ids)
```

```python
# [16]
encoded = tokenizer(text)
print('encoded:', encoded)
```

```python
# [17]
print('Input IDs:', encoded['input_ids'])
```

```python
# [18]
# "Hi CeraGPT"           #=> [1, 6324, 2,    2,   2]   // length 5
# "Is SPF good daily?"   #=> [1, 1317, 10937, 1781, 29973]  // length 5
#
# "Hi CeraGPT" uchun      'attention_mask': [1, 1, 0, 0, 0]
# "Is SPF good daily?" uchun 'attention_mask': [1, 1, 1, 1, 1]
```

```python
# [19]
decoded_text = tokenizer.decode(encoded["input_ids"])
print('decoded_text:', decoded_text)
```

```python
# [29]
texts = [
    "Hello, CeraGPT!",
    "Today I want to find a good sunscreen.",
    "CeraGPT is a small but smart skincare assistant."
]

batch = tokenizer(texts, padding=True, return_tensors="pt")
print(batch.keys())

print("Input IDs shape:", batch["input_ids"].shape)
print("Attention mask shape:", batch["attention_mask"].shape)
```
